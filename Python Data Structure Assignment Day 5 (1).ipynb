### Create colors = [‘red’,’blue’,’orange’], and fruits = [‘orange’,’grapes’,’apples’] 

```python
colors = ["red", "blue", "orange"]
fruits = ["orange", "grapes", "apples"]
```

Print color and fruits

```python

print("Colors list:", colors)
print("Fruits list:", fruits)
```

Create colors_set, and fruits_set. (using set() ) and print them

```python
colors_set = set(colors)
fruits_set = set(fruits)

print("Colors set:", colors_set)
print("Fruits set:", fruits_set)
```

Find the union of both the sets.

```python

union_set = colors_set.union(fruits_set)
print("Union of colors and fruits:", union_set)
```

Find the intersection of both the sets 

```python
intersection_set = colors_set.intersection(fruits_set)
print("Intersection of colors and fruits:", intersection_set)

```

Find the elements which are Fruits but not colors (using set.difference() )

```python
difference_set = fruits_set.difference(colors_set)
print("Fruits but not colors:", difference_set)

```

```python
union_set = colors_set.union(fruits_set)
print("Union of colors and fruits:", union_set)

intersection_set = colors_set.intersection(fruits_set)
print("Intersection of colors and fruits:", intersection_set)

difference_set = fruits_set.difference(colors_set)
print("Fruits but not colors:", difference_set)
```

count the number of occurences of charachters in string named "sent" using dictionary and print the same.


```python
sent = "hip hop album still here Kr$na"
char_count = {}

for ch in sent:
    if ch in char_count:
        char_count[ch] += 1
    else:
        char_count[ch] = 1

print(char_count)
```

```python
for key, value in char_count.items():
    print(f"{key} ----> {value}")
```
